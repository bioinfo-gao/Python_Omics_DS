Markdown: 我们将在 Markdown (.md) 文件中编写文档。

第二步：学习 LaTeX 公式的基本语法
LaTeX 公式有两种主要的插入模式：行内公式 (Inline) 和块级公式 (Block/Display)。

1. 行内公式 (Inline Formula)
作用：将简短的公式嵌入到普通文本行中。
语法：用一对美元符号 $...$ 将公式包裹起来。

当我们讨论一个变量 x 的平方时，我们写作 $x^2$。一个著名的公式是爱因斯坦的质能方程 $E=mc^2$。

2. 块级公式 (Display Formula)
作用：将复杂的、需要单独突出显示的公式放在一个独立的行中，并居中显示。
语法：用两对美元符号 $$..   .$$ 将公式包裹起来。

### Sigmoid 函数 (Sigmoid Function)

Sigmoid 函数是一个在机器学习中常用的激活函数，它的值域在 (0, 1) 之间。其数学公式如下：

$$
\sigma(x) = \frac{1}{1 + e^{-x}}
$$

这个函数可以将任意实数输入映射到 (0, 1) 的输出，非常适合用于表示概率。
