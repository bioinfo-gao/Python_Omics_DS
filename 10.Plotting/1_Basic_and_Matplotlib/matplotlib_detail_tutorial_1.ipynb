5.用饼图的面积及颜色展示一组4维数据

```python
# 创建一个伪随机数生成器的实例，并设置种子为0。
rng = np.random.RandomState(0)

# 生成一个包含100个随机数的一维数组，这些数服从标准正态分布（均值为0，方差为1）。
x = rng.randn(100)
y = rng.randn(100)

colors = rng.rand(100) 
# plt.scatter()函数接收到这100个数值后，会使用cmap='viridis'这个颜色映射表（colormap）
# 将这些数值转换成具体的颜色。

sizes = 1000 * rng.rand(100)
# 将每个数乘以1000。这些结果值将决定每个点的大小。

# 调用scatter函数来绘制散点图。
# x: 指定所有点的x坐标。
# y: 指定所有点的y坐标。
# c=colors: 将colors数组中的值传递给参数c，用于指定每个点的颜色。matplotlib会根据cmap将这些数值映射成具体颜色。
# s=sizes: 将sizes数组中的值传递给参数s，用于指定每个点的面积大小。
# alpha=0.3: 设置所有点的透明度为30%（0.3），这样可以更容易地看到重叠的点。
# cmap='viridis': 指定使用的颜色映射表（colormap）为'viridis'。它会把colors数组中的数值（从0到1）映射为从紫色到黄色的渐变色。
plt.scatter(x, y, c=colors, s=sizes, alpha=0.3, cmap='viridis') # <<==============

# 在图表的旁边添加一个颜色条（colorbar）。
# 这个颜色条会显示数值（来自colors数组）与实际颜色之间的对应关系。
plt.colorbar()
#  当你调用 plt.colorbar() 函数时，它会在你的图表上绘制一个颜色条。
# 同时，这个函数会创建并返回一个 Colorbar 对象。这个对象包含了颜色条的所有信息和属性。

# plt.colorbar(); 
# # very interesting, with ; the matplotlib,object will show up in jupyter notebook
# 分号 (;) 可以用来在同一行中分隔多条语句。然而，在Jupyter和IPython这样的环境中，它还有一个特殊的、更常见的用途：抑制当前行代码的输出。

# （可选，但在脚本中通常是必需的）显示生成的图形。
# plt.show()
```

在进行数据可视化时，我们通常只关心生成的图像，而不关心绘图函数返回的对象本身。因此，在Jupyter环境中，为了保持输出的整洁，在绘图命令（尤其是单元格的最后一行）的末尾加上分号是一种非常普遍且推荐的做法。

23.在一张图里绘制sin,cos的图形,并展示图例

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 10, 1000)

# 使用 matplotlib.pyplot.subplots() 创建一个新的图形 (fig) 和一个或多个子图/坐标系 (ax)。
# fig 是整个窗口，ax 是实际绘图的区域。这是推荐的面向对象绘图方式。
fig, ax = plt.subplots()

# label='sin' 为这条曲线设置图例标签。
ax.plot(x, np.sin(x), label='sin')

# 在坐标系 ax 上绘制第二条曲线：x 轴为 x 数组，y 轴为 x 数组中每个值的余弦 (cos) 值。
# '-' 指定了线的样式为实线（这是默认值，但显式指定也可）。label='cos' 设置图例标签。
ax.plot(x, np.cos(x), '-', label='cos')

# 调用 ax.legend() 方法，根据 ax.plot() 中设置的 label 来显示图例。
ax.legend(); 


# 注意：在实际的 Python 脚本中，如果要在 Jupyter 或交互式环境之外看到图形，
# 通常还需要添加 plt.show() 命令。
```

26.绘制图像,只显示2个图例  sin(x), sin(x + pi / 2), sin(x + pi)  sin(x + 3/2 * pi) ==>>>>> 4条线但只有两个图例

```python
# 变量 x 继承自上一个代码块，它是一个包含 1000 个从 0 到 10 的等间距点的 NumPy 数组（形状为 (1000,)）。

# 使用 NumPy 广播机制生成一个多维数组 y。
# 具体步骤如下：    
# 1. x[:, np.newaxis]：将 x 数组的形状从 (1000,) 变为 (1000, 1)。    << ============ 一维变二维
# 2. np.arange(0, 2, 0.5)：创建一个包含 [0.0, 0.5, 1.0, 1.5] 的数组（长度为 4）。
# 3. np.pi * ...：将上一步的数组乘以 pi。
# 4. x[:, np.newaxis] + np.pi * ...：进行广播加法。结果是一个形状为 (1000, 4) 的数组，
#    其中每一列都是一个正弦波，相位依次增加 0, 0.5*pi, 1.0*pi, 1.5*pi。
# 5. y = np.sin(...)：计算最终的 y 值。y 的形状是 (1000, 4)。
y = np.sin(x[:, np.newaxis] + np.pi * np.arange(0, 2, 0.5))
print(y.shape)  # 输出 (1000, 4)
print(y[:5, :])  # 输出前5行数据以查看结果 注意第一行： 0，  1，   0 ， -1 
```

```python

# 绘制曲线。由于 y 是一个 (1000, 4) 的二维数组，plt.plot(x, y) 会自动绘制 4 条曲线。
# lines 变量将捕获这 4 条曲线对应的 plt.Line2D 对象（一个包含 4 个元素的列表）。
lines = plt.plot(x, y)

# Lines 是 plt.Line2D 类型的实例的列表

# 自定义图例。plt.legend() 可以接受两个参数：
# 1. lines[2]：表示要显示图例的 Line2D 对象的列表。这里只选择列表中的第三个对象（索引 2），即第三条曲线。
# 2. ['first', 'second']：表示为这些对象指定的标签（labels）。
# 
# **注意：** 传入的 Line2D 对象数量与标签数量必须匹配。
# **这里有一个潜在的错误：** lines[2] 只是**一个** Line2D 对象，但标签列表有**两个**元素 (['first', 'second'])。
# 除非 lines[2] 是一个列表，否则代码会报错，或者 Matplotlib 可能会只显示第一个标签并忽略第二个，具体行为取决于 Matplotlib 版本和环境。
# 假设目的是为了演示图例自定义，但该调用可能需要修正为：
# plt.legend(lines[0:2], ['first', 'second']); # 选择前两条线并分别标记为 'first', 'second'
# 或者
# plt.legend([lines[2]], ['third line']); # 正确标记第三条线
# 但是，根据原代码，它的字面意图是：使用第三条线作为图例的**手柄**，并尝试用 'first', 'second' 来标记它。
plt.legend(lines[:2], ['first', 'second']);
```

3.11 以后的新方法

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors # 导入 colors 模块

# 假设 I 是您的图像数据
# I = np.random.rand(10, 10) 
# 为了演示，创建一个示例图像，包含 0 到 5 的整数值，以匹配 6 个分段
# I = np.random.randint(0, 6, size=(10, 10)) 

I = np.sin(x) * np.cos(x[:, np.newaxis])
# -------------------------- 新方法 --------------------------
N_colors = 6 # 您想要的离散颜色数量

# 1. 获取基础颜色映射
cmap_base = plt.colormaps['Blues']

# 2. 定义边界：N 个颜色需要 N+1 个边界。
#    这里我们假设数据范围从 0 到 6 (即 0, 1, 2, 3, 4, 5, 6)
# bounds = np.arange(N_colors + 1) - 0.5 
# 例如，对于 N=6，bounds 将是 [-0.5, 0.5, 1.5, 2.5, 3.5, 4.5, 5.5]，这能完美地将整数数据 0-5 分段


# 2. 定义边界：创建 N+1 个边界，从 -1.0 到 1.0 均匀分布
#    np.linspace(-1, 1, N_colors + 1) 会生成 7 个边界点：
#    [-1.0, -0.666..., -0.333..., 0.0, 0.333..., 0.666..., 1.0]
bounds = np.linspace(-1.0, 1.0, N_colors + 1)


# 3. 创建归一化对象
#    它将数据 I 映射到 N 个分段，并使用 cmap_base 中的 N 种颜色
norm = colors.BoundaryNorm(bounds, cmap_base.N) 

# 4. 绘制图像
plt.imshow(I, cmap=cmap_base, norm=norm)

# 可选：添加颜色条来显示分段
##plt.colorbar(ticks=np.arange(N_colors)) 
plt.colorbar() 
plt.clim(-1, 1);
plt.show()
```
