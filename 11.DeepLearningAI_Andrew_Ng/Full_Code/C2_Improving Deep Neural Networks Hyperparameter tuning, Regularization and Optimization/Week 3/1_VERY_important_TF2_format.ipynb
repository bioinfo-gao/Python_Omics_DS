| 功能模块 | 旧版语法 (TF 1.x) | 新版适配 (TF 2.x) | 3060 显卡适配说明与注意事项 |
| :--- | :--- | :--- | :--- |
| **执行模式** | 静态图 (Static Graph) | **动态图 (Eager Execution)** | TF2 默认开启，代码写完即可运行并查看结果，无需等待。 |
| **数据入口** | `tf.placeholder()` | **直接传入 Tensor / Array** | 彻底废弃。在 R44 环境下直接将数据作为参数传给函数即可。 |
| **运行引擎** | `tf.Session().run()` | **直接调用函数** | 彻底废弃 Session。像运行普通 Python 脚本一样直接执行。 |
| **变量管理** | `tf.get_variable()` | **`tf.Variable()`** | 废弃变量作用域（Scope）。直接创建变量对象，支持即时初始化。 |
| **梯度计算** | `optimizer.minimize()` | **`tf.GradientTape()`** | **核心差异**：必须手动记录前向传播过程来求导，灵活性更高。 |
| **参数更新** | 自动在 Session 中更新 | **`optimizer.apply_gradients`** | 配合 Adam 等优化器，显式应用计算出的梯度值。 |
| **矩阵运算** | `tf.matmul`, `tf.add` | **`tf.linalg.matmul` / 算术符** | TF2 支持 `+`, `-`, `*` 直接对 Tensor 进行运算。 |
| **图像读取** | `scipy.ndimage.imread` | **`plt.imread`** | Scipy 1.2+ 已移除此功能，Matplotlib 是 Windows 11 下更稳妥的选择。 |
| **图像缩放** | `scipy.misc.imresize` | **`PIL.Image.resize`** | 推荐使用 Pillow 库，效率更高且符合 Python 3.10 标准。 |
| **显存清理** | `tf.reset_default_graph()` | **重启 Notebook Kernel** | TF2 显存占用是动态的，若 3060 报 OOM 错误需手动重启内核。 |

最后的锦囊妙计
显存回收：如果你开始进行更复杂的实验（比如训练几千张图），Jupyter 可能不会自动释放显存。如果 3060 报错 OOM (Out of Memory)，请点击菜单栏的 Kernel -> Restart，这是最有效的物理清空方法。

数据类型：记得永远优先使用 tf.cast(data, tf.float32)。你的显卡核心对 float32 有专门的硬件加速（Tensor Cores），这会让你的模型训练快得多。
