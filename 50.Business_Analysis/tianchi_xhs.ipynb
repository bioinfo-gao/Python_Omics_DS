基于python+Power BI的淘宝用户行为分析
项目介绍:本项目是基于2022年BI数据分析大赛,通过一个淘宝用户数据集对用户购物的行为进行数据分析,深度挖掘出用户购物行为的参考依据、用户行为倾向以及用户流失分析。


一、数据预处理
1.数据来源
数据集:
tianchi_mobile_recommend_train_item.csv
tianchi_mobile_recommend_train_user.zip

1.原数据字段

user_id 表示的是用户id。
item id 表示商品id。
user geohash 表示用户地址。
item_category表示商品类别。
behavior_type 表示的是用户行为,包括点击、收藏、加购、购买,分别用1,2,3,4字段表示。
time 表示用户行为发生时间,包括日期和小时。

1.数据预处理
(1)数据去重
利用pandas库先读取文件,并反馈前几行数据以及数据总数;并利用 df.drop_duplicates()对各个字段同时满足重复的重复数据进行去除,并再次反馈数据的前几行以及重复数据。
