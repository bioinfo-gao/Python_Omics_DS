```python
# 文件：shipbuilding_market_share.py
# 运行：pip install matplotlib pandas

import matplotlib.pyplot as plt
import pandas as pd

# 模拟数据（基于文章）
data = pd.DataFrame({
    'Country': ['China', 'US', 'South Korea', 'Japan'],
    'Market Share (%)': [60, 0.1, 20, 15]  # 2024年估算
})

# 绘制饼图
plt.figure(figsize=(8, 6))
plt.pie(data['Market Share (%)'], labels=data['Country'], autopct='%1.1f%%', startangle=90)
plt.title('Global Large Vessel Shipbuilding Market Share (2024)')
plt.show()

# 保存图表
plt.savefig('shipbuilding_share.png')
```
