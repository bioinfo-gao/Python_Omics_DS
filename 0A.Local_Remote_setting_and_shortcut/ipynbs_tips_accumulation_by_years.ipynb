#### 1 如何复制、剪切、粘贴多个cell
　
步骤一：选择起始
点击 您想要开始剪切的那个cell 的左侧边栏。此时，该cell会被选中（左侧边栏会变蓝或变绿）。

步骤二：选中后续所有 
从您需要选中这个起始 鼠标向下滚动到想选择最后一个cell , 不用点击或者压住任何键，单纯滚动就行。
按住键盘的Shift键。并保持按住
此时点击最后一个cell的左侧边栏。
这样，从起始到末尾的所有cell都会被一次性选中。

然后可以Ctrl+C复制选中的cell。
然后可以Ctrl+X剪切选中的cell。
然后可以Ctrl+V粘贴选中的cell。
